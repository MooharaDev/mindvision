# pdf2db output inspector

Shows the top 5 rows of every CSV artifact produced by `pdf2db.py`.

Set `OUT_DIR` below to the folder you passed as `--out`, then run all cells.

```python
from pathlib import Path
import pandas as pd

OUT_DIR = Path("test_out")  # <-- change to your --out folder if different

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 60)
```

```python
from IPython.display import display, Markdown

csv_paths = sorted(OUT_DIR.glob("*.csv"))
if not csv_paths:
    print(f"No CSV files found in {OUT_DIR.resolve()}\n"
          f"Run pdf2db.py first, or point OUT_DIR at the right folder.")

for path in csv_paths:
    df = pd.read_csv(path)
    display(Markdown(f"### `{path.name}` \u2014 {df.shape[0]} rows \u00d7 {df.shape[1]} columns"))
    display(df.head(5))
```

**Reading guide:** start with `records.csv` (one row per record: extracted fields, `_confidence`, `status`, `review_reasons`), then `review_queue.csv` (what needs human eyes) and `issues.csv` (every skipped/suspicious file). `pdfs.csv` / `text_report.csv` show per-PDF text yield \u2014 `low_yield=1` means probable scan.
